#Popularity based recommendation system
#I will consider only the books that have got votes more than 250
#Then I will show the top 50 books means books with highest top 50 average rating


```python
rating_with_books=ratings.merge(books,on='ISBN')
rating_with_books.drop_duplicates(inplace=True)
rating_with_books.head()
```

#Let's find out the books that have number of book_ratings more than 250

#Book Titles based on number of ratings

```python
number_of_ratings=rating_with_books.groupby('Book-Title').count()['Book-Rating'].reset_index()
```

```python
number_of_ratings.rename(columns={'Book-Rating':'num_rating'},inplace=True)
```

```python
number_of_ratings.head(50)
```

#Book titles based on average rating

```python
rating_with_books['Book-Rating'] = pd.to_numeric(rating_with_books['Book-Rating'], errors='coerce')
avg_ratings=rating_with_books.groupby('Book-Title').mean(numeric_only=True)['Book-Rating'].reset_index()
avg_ratings.rename(columns={'Book-Rating':'avg_rating'},inplace=True)
avg_ratings.head()
```

```python
popular_df=number_of_ratings.merge(avg_ratings,on='Book-Title')
```

```python
popular_df.head()
```

```python
popular_df=popular_df[popular_df['num_rating']>=250].sort_values('avg_rating',ascending=False).head(50)
popular_df.head()
```

```python
popular_df.shape
```

#Now to get the book data I need to merge with books df again

```python
popular_df=popular_df.merge(books,on='Book-Title').drop_duplicates('Book-Title')[['Book-Title','Book-Author','Image-URL-M','num_rating','avg_rating']]
```

#Now starting the collaborative filtering based recommendation System


```python
active_users=rating_with_books.groupby('User-ID').count()['Book-Rating']>200
active_users=active_users[active_users].index
active_users
```

```python
filtered_rating=rating_with_books[rating_with_books['User-ID'].isin(active_users)]
```

```python
filtered_rating.head()
```

```python
famous_books=rating_with_books.groupby('Book-Title').count()['Book-Rating']>=50
famous_books=famous_books[famous_books].index
famous_books
```

```python
filter_rating=filtered_rating[filtered_rating['Book-Title'].isin(famous_books)]
filter_rating.head()
filter_rating.shape
```

```python
filter_rating=filter_rating.drop_duplicates()
```

```python
filter_rating.head()
```

```python
pt=filter_rating.pivot_table(index='Book-Title',columns='User-ID',values='Book-Rating')
pt.fillna(0,inplace=True)
pt.shape
```

```python
pt
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity=cosine_similarity(pt)
similarity.shape
```

```python
np.where(pt.index=='1984')[0][0]
```

```python
sorted(list(enumerate(similarity[0])),key=lambda x:x[1],reverse=True)
```

```python
def recommend(book_name):
  book_index=np.where(pt.index==book_name)[0][0]
  distances=similarity[book_index]
  similar_items=sorted(list(enumerate(distances)),key=lambda x:x[1],reverse=True)[1:10]

  for i in similar_items:
    print(pt.index[i[0]])
```

```python
recommend('1984')
```

```python
import pickle
pickle.dump(filter_rating,open('filter_rating.pkl','wb'))
pickle.dump(pt,open('pt.pkl','wb'))
pickle.dump(similarity,open('similarity.pkl','wb'))
```
